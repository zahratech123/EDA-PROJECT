```python
correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
corr_matrix = df[numeric_columns].corr()

for column in corr_matrix.columns:
    for row in corr_matrix.index:
        if column != row and abs(corr_matrix.loc[row, column]) >= 0.7:
            print(
                row,
                "and",
                column,
                "correlation =",
                round(corr_matrix.loc[row, column], 2)
            )
```
